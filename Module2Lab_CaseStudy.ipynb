```python
# Dhanashri Sisodiya
# Modul 2 Case Study
# This program checks student's GPAs for Dean's List and Honor Roll.

last_name = input("Enter student's last name (ZZZ to quit): ")

while last_name != "ZZZ":
    first_name = input("Enter student's first name: ")
    gpa = float(input("Enter student's GPA"))

    if gpa >= 3.5:
        print(first_name, last_name, "has made the Dean's List.")
    elif gpa >= 3.25:
        print(first_name, last_name, "had made to the Honor Roll.")
    else:
        print(first_name, last_name, "does not qualift for either list.")
    
    last_name = input("Enter student's last name (ZZZ to quit)")


    
```
