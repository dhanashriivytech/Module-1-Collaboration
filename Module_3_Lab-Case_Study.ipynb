```python
# Student Name: Dhanashri Sisodiya
# File Name: Module_3_Lab-Case_Study.ipynb
# Course: SDEV 220 - Python
# Assignment: Module 3 Lab - Case Study: Lists, Functions, and Classes
#
# Description:
# This program uses a Vehicle superclass and an
# Automobile subclass to collect and display car
# information entered by the user.
#
# Variables:
# vehicle_type : It stores the type of the vehicle.
# year : It stores the automobile's model year.
# make : It stores the automobile manufacturer.
# model : It stores the automobile model name.
# doors : It stores the number of doors (2 or 4).
# roof : It Stores the roof type (solid or sun roof).
# car : It stores the Automobile object.
```

```python
# Create super class called Vehicl
class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type


# Create the Automobile class
class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof


# Ask the user to enter car information
year = input("Enter the year: ")
make = input("Enter the make: ")
model = input("Enter the model: ")
doors = input("Enter the number of doors (2 or 4): ")
roof = input("Enter the type of roof (solid or sun roof): ")


# Create an Automobile object
car = Automobile(year, make, model, doors, roof)


# Display the car information
print()
print("Vehicle Information")
print("...................")
print("Vehicle type:", car.vehicle_type)
print("Year:", car.year)
print("Make:", car.make)
print("Model:", car.model)
print("Number of doors:", car.doors)
print("Type of roof:", car.roof)
```
