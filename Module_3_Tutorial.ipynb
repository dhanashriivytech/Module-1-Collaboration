```python

#  Sort an Array of 0s, 1s, and 2s
class Solution:
    def sort012(self, arr):
        count0 = 0
        count1 = 0
        count2 = 0

        for num in arr:
            if num == 0:
                count0 += 1
            elif num == 1:
                count1 += 1
            else:
                count2 += 1

        i = 0

        for x in range(count0):
            arr[i] = 0
            i += 1

        for x in range(count1):
            arr[i] = 1
            i += 1

        for x in range(count2):
            arr[i] = 2
            i += 1

        return arr


# Pass the values to the program
numbers = [0, 1, 2, 0, 1, 2]

obj = Solution()

print("Original List:", numbers)
print("Sorted List:", obj.sort012(numbers))
```

```python
#  Binary Search
# Description: Find the first occurrence of a number in a sorted list.

class Solution:
    def firstSearch(self, arr, k):
        start = 0
        end = len(arr) - 1
        result = -1

        while start <= end:
            mid = (start + end) // 2

            if arr[mid] == k:
                result = mid
                end = mid - 1

            elif arr[mid] < k:
                start = mid + 1

            else:
                end = mid - 1

        return result


# Pass values to the program
arr = [1, 2, 3, 4, 5]
k = 4

obj = Solution()
answer = obj.firstSearch(arr, k)

print("First occurrence at index:", answer)
```
