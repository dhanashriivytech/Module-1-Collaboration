```python
# Exercise 16.4

import sqlite3

conn = sqlite3.connect('books.db')
cursor = conn.cursor()

cursor.execute('''CREATE TABLE IF NOT EXISTS books
                  (title TEXT, author TEXT, year INTEGER)''')

conn.commit()
conn.close()

print('Books table created successfully')
```

```python

# Exercise 16.5

import csv
import sqlite3

conn = sqlite3.connect('books.db')
cursor = conn.cursor()

with open('books2.csv') as file:
    books = csv.DictReader(file)

    for book in books:
        cursor.execute('INSERT INTO books VALUES (:title, :author, :year)', book)

conn.commit()
conn.close()
```

```python
# Exercise 16.8

import sqlalchemy

engine = sqlalchemy.create_engine('sqlite:///books.db')
conn = engine.connect()

result = conn.execute(sqlalchemy.text('SELECT title FROM books ORDER BY title'))

for book in result:
    print(book[0])

conn.close()
```
